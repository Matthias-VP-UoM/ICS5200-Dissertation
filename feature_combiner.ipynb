```python
import pandas as pd
import os
```

```python
# Load the feature sets
dom = pd.read_csv("data/features/dom_features.csv")
img = pd.read_csv("data/features/image_features.csv")
# dom = pd.read_csv("data/features/dom_features_new_v5.csv")
# img = pd.read_csv("data/features/image_features_v2.csv")

# Load the manifest file (adjust the path to match your directory structure if needed)
manifest = pd.read_csv("manifest.csv") 

# Merge the DOM and image features
final = dom.merge(img, on=["id", "website", "page_type"])

# Merge the manifest columns using the common identifier keys
final = final.merge(
    manifest[["id", "website", "page_type", "screenshot_path", "html_path", "url"]], 
    on=["id", "website", "page_type"]
)

# 1. Define your primary columns (including the new metadata columns from the manifest)
front_cols = ["id", "website", "page_type", "screenshot_path", "html_path", "url"]

# 2. Get all other columns that aren't in the front list
other_cols = [col for col in final.columns if col not in front_cols]

# 3. Combine them and reindex the DataFrame
final = final[front_cols + other_cols]

# Save the updated dataset
final.to_csv("data/final_dataset.csv", index=False)
# final.to_csv("data/final_dataset_v6.csv", index=False)
```
